```python
# Define file paths and content
sample_texts = {
    "../data/text_files/ml_intro.txt" :
        """Machine Learning Overview

1. Definition:
   Machine Learning (ML) is a subset of Artificial Intelligence (AI) that enables systems to learn patterns from data and improve performance without explicit programming.

2. Types of ML:
   - Supervised Learning: Models trained on labeled data (e.g., regression, classification).
   - Unsupervised Learning: Models discover hidden patterns in unlabeled data (e.g., clustering, dimensionality reduction).
   - Reinforcement Learning: Agents learn by interacting with an environment using rewards and penalties.

3. Key Concepts:
   - Dataset: Collection of examples used for training/testing.
   - Features: Input variables describing data.
   - Labels: Target outputs in supervised learning.
   - Model: Mathematical representation mapping inputs to outputs.
   - Training: Process of fitting the model to data.
   - Evaluation: Measuring model performance using metrics (accuracy, precision, recall, F1-score).

4. Common Algorithms:
   - Linear Regression
   - Logistic Regression
   - Decision Trees
   - Random Forests
   - Support Vector Machines (SVM)
   - k-Nearest Neighbors (k-NN)
   - Neural Networks

5. Applications:
   - Image recognition
   - Natural language processing (NLP)
   - Fraud detection
   - Recommendation systems
   - Predictive analytics

6. Challenges:
   - Overfitting vs. underfitting
   - Data quality and preprocessing
   - Model interpretability
   - Scalability and deployment

Summary:
Machine Learning is about building systems that learn from data. By mastering algorithms, evaluation techniques, and deployment strategies, ML engineers can create solutions that solve real-world problems across industries.
""",

    "../data/text_files/rag_intro.txt" :
        """Retrieval-Augmented Generation (RAG) Overview

1. Definition:
   Retrieval-Augmented Generation (RAG) is an advanced AI technique that combines large language models (LLMs) with external knowledge retrieval systems. It enables models to generate responses grounded in up-to-date, domain-specific information rather than relying solely on pre-trained parameters.

2. Core Idea:
   - Retrieval: Fetch relevant documents or data from an external source (e.g., vector database, search engine).
   - Generation: Use the retrieved context to produce accurate, coherent, and context-aware responses.

3. Key Components:
   - Embeddings: Numerical representations of text used for similarity search.
   - Vector Database: Stores embeddings for efficient retrieval (e.g., FAISS, Pinecone, Weaviate).
   - Retriever: Finds the most relevant context for a query.
   - Generator: LLM that produces the final answer using both query and retrieved context.

4. Workflow:
   - User query → Convert to embedding → Retrieve top‑k relevant documents → Pass query + documents to LLM → Generate grounded response.

5. Applications:
   - Enterprise chatbots with company knowledge bases
   - Question answering systems
   - Legal, medical, and financial assistants
   - Research tools for summarizing large document sets
   - Customer support automation

6. Advantages:
   - Provides factual, up-to-date answers
   - Reduces hallucinations in LLM outputs
   - Scales to large knowledge bases
   - Flexible integration with APIs and databases

7. Challenges:
   - Quality of retrieval affects accuracy
   - Latency due to retrieval + generation steps
   - Requires careful system design (indexing, caching, reranking)
   - Balancing relevance vs. diversity of retrieved documents

Summary:
RAG bridges the gap between static language models and dynamic knowledge sources. By combining retrieval with generation, it creates AI systems that are both intelligent and informed, making them highly valuable for real-world applications.
"""
}

# Write files
for filepath, content in sample_texts.items():
    with open(filepath, "w", encoding="utf-8") as f:
        f.write(content)

print("Sample file texts created!!")
```

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/text_files/ml_intro.txt",encoding="utf-8")
document = loader.load()

import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)

print(document)
```

```python
from langchain_community.document_loaders import DirectoryLoader, TextLoader

dir_loader = DirectoryLoader(
    "../data/text_files/",
    glob="**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={"encoding": "utf-8"},
    show_progress=False
)
documents = dir_loader.load()
print(documents)
```

```python
from langchain_community.document_loaders import DirectoryLoader, PyMuPDFLoader

dir_loader = DirectoryLoader(
    "../data/pdf/",
    glob="**/*.pdf",
    loader_cls=PyMuPDFLoader
)

pdf_documents = dir_loader.load()
print(pdf_documents)
```
